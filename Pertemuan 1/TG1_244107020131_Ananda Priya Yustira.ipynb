## Fungsi Library

### 1. PyPREP
PyPREP digunakan untuk melakukan preprocessing pada sinyal EEG (Electroencephalography). Library ini membantu membersihkan data EEG, mendeteksi channel yang bermasalah atau memiliki banyak noise, melakukan filtering, serta robust referencing agar sinyal lebih siap untuk dianalisis lebih lanjut.

### 2. SciPy
SciPy adalah library Python yang digunakan untuk scientific computing. Library ini menyediakan berbagai fungsi untuk perhitungan matematika, statistik, optimasi, serta pengolahan sinyal. Dalam analisis sinyal seperti EEG dan ECG, SciPy dapat digunakan untuk filtering, transformasi Fourier, analisis frekuensi, dan proses signal processing lainnya.

### 3. Weights & Biases (wandb)
Weights & Biases atau `wandb` digunakan untuk mencatat, memonitor, dan membandingkan eksperimen Machine Learning. Library ini dapat menyimpan informasi seperti nilai loss, accuracy, hyperparameter, jumlah epoch, serta konfigurasi model. Hasil eksperimen juga dapat divisualisasikan melalui dashboard agar proses evaluasi model lebih mudah dilakukan.

### 4. pyECG
pyECG digunakan untuk pemrosesan dan analisis sinyal Electrocardiogram (ECG). Library ini membantu dalam menganalisis aktivitas listrik jantung, seperti mendeteksi denyut jantung, R-peak, interval antar denyut, serta mengekstraksi informasi penting dari sinyal ECG.
