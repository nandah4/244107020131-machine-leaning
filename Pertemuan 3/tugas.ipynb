# Praktikum 3 - Preprocessing dan Stratified Split Wisconsin Breast Cancer

Dataset **Wisconsin Breast Cancer** terdiri dari 569 data yang digunakan untuk mendiagnosis kanker menjadi dua kelas, yaitu **Malignant (M)** dan **Benign (B)**.

Pada praktikum ini dilakukan beberapa proses:
1. Memisahkan variabel yang dapat dan tidak dapat digunakan.
2. Melakukan encoding pada kolom `diagnosis`.
3. Melakukan standardisasi pada variabel numerik.
4. Membagi data menjadi data latih dan data uji menggunakan *stratified split* dengan rasio 80:20.

```python
import pandas as pd

df = pd.read_csv("wbc.csv")

df.head()
```

## 1. Memisahkan Variabel yang Dapat dan Tidak Dapat Digunakan

Pada dataset terdapat beberapa variabel yang tidak diperlukan dalam proses pengolahan data:

- `id` tidak digunakan karena hanya berfungsi sebagai identitas unik setiap data.
- `Unnamed: 32` tidak digunakan karena merupakan kolom kosong.
- `diagnosis` digunakan sebagai variabel target.
- Kolom lainnya digunakan sebagai variabel fitur karena berisi karakteristik numerik hasil pemeriksaan.

```python
# Menghapus kolom yang tidak digunakan
df = df.drop(columns=['id', 'Unnamed: 32'])

df.head()
```

Selanjutnya data dipisahkan menjadi:

- **X (fitur)**: seluruh variabel karakteristik tumor.
- **y (target)**: variabel `diagnosis`.

```python
X = df.drop(columns=['diagnosis'])
y = df['diagnosis']

print("Jumlah fitur:", X.shape[1])
print("Jumlah data:", X.shape[0])
```

## 2. Encoding Kolom Diagnosis

Kolom `diagnosis` masih berupa data kategorikal:

- `B` = Benign
- `M` = Malignant

Oleh karena itu, dilakukan proses encoding menggunakan `LabelEncoder` agar nilai kategorikal tersebut berubah menjadi nilai numerik.

```python
from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()

y = encoder.fit_transform(y)

print("Kategori:", encoder.classes_)
print("Hasil encoding:", y[:10])
```

Hasil encoding pada variabel `diagnosis` adalah:

- **Benign (B) = 0**
- **Malignant (M) = 1**

Dengan demikian, variabel target sudah berbentuk numerik dan dapat digunakan dalam proses selanjutnya.

## 3. Standardisasi Data Numerik

Setiap fitur pada dataset memiliki skala nilai yang berbeda. Oleh karena itu, dilakukan standardisasi menggunakan `StandardScaler` agar setiap fitur memiliki rata-rata mendekati 0 dan standar deviasi mendekati 1.

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)
```

Hasil standardisasi kemudian dikembalikan ke dalam bentuk DataFrame agar lebih mudah dibaca.

```python
X_scaled = pd.DataFrame(X_scaled, columns=X.columns)

X_scaled.head()
```

## 4. Stratified Split Data

Data kemudian dibagi menjadi **data latih (training)** dan **data uji (testing)** dengan rasio **80:20**.

Proses pembagian menggunakan metode *stratified split*. Metode ini digunakan agar proporsi kelas `Benign` dan `Malignant` pada data latih dan data uji tetap menyerupai proporsi pada dataset awal.

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_scaled,
    y,
    test_size=0.2,
    random_state=42,
    stratify=y
)
```

Setelah proses pembagian, dilakukan pengecekan jumlah data latih dan data uji.

```python
print("Jumlah data training:", X_train.shape[0])
print("Jumlah data testing :", X_test.shape[0])
```

Untuk memastikan proses *stratified split* berhasil, dilakukan pengecekan proporsi kelas pada data asli, data training, dan data testing.

```python
print("Proporsi data asli:")
print(pd.Series(y).value_counts(normalize=True))

print("\nProporsi data training:")
print(pd.Series(y_train).value_counts(normalize=True))

print("\nProporsi data testing:")
print(pd.Series(y_test).value_counts(normalize=True))
```
