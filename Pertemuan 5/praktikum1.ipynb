# Pertemuan 5 — Praktikum 1: K-Means pada Data Iris

Tujuan praktikum ini adalah mengelompokkan bunga Iris berdasarkan ukuran sepal dan petal. Sepal adalah kelopak, sedangkan petal adalah mahkota bunga.

Sumber: [Praktikum 1](https://polinema.gitbook.io/jti-modul-praktikum-pembelajaran-mesin-2025-2026/js04-klasterisasi/praktikum-1). Dataset `Iris.csv` diunduh dari lampiran modul.


## 1. Membaca data
Kita membaca dataset dan memeriksa jumlah data serta nilai yang kosong.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

df = pd.read_csv('Iris.csv')
print('Ukuran data:', df.shape)
print('Jumlah nilai kosong:', df.isna().sum().sum())
df.head()
```

## 2. Memilih fitur
Kolom `Id` tidak dipakai karena hanya nomor urut. Kolom `Species` juga tidak digunakan untuk melatih model karena K-Means mengelompokkan data tanpa label asli. Empat kolom ukuran bunga menjadi fitur model.


```python
X = df[['SepalLengthCm', 'SepalWidthCm', 'PetalLengthCm', 'PetalWidthCm']]
y = df['Species']  # Hanya untuk melihat jenis bunga, bukan untuk melatih model.
X.head()
```

## 3. Melihat sebaran data
Model memakai empat fitur, tetapi grafik hanya menampilkan dua fitur agar mudah dilihat.


```python
plt.figure(figsize=(7, 5))
plt.scatter(X['SepalLengthCm'], X['SepalWidthCm'], s=50)
plt.xlabel('Panjang sepal (cm)')
plt.ylabel('Lebar sepal (cm)')
plt.title('Sebaran Data Iris')
plt.show()
```

## 4. Membuat model dengan k = 2


```python
cl_kmeans = KMeans(n_clusters=2, random_state=0, n_init=10)
y_kmeans = cl_kmeans.fit_predict(X)
centers = cl_kmeans.cluster_centers_

plt.figure(figsize=(7, 5))
plt.scatter(X.iloc[:, 0], X.iloc[:, 1], c=y_kmeans, s=50, cmap='viridis')
plt.scatter(centers[:, 0], centers[:, 1], c='red', s=200, marker='X', label='Centroid')
plt.xlabel('Panjang sepal (cm)')
plt.ylabel('Lebar sepal (cm)')
plt.title('K-Means Iris dengan k = 2')
plt.legend()
plt.show()
print(f'Nilai SSE: {cl_kmeans.inertia_:.3f}')
```

## 5. Mencari jumlah klaster dengan metode Elbow
SSE adalah jumlah kuadrat jarak data terhadap centroid terdekat. Kita mencoba k dari 1 sampai 10. Pilihan k dilihat pada bagian grafik yang mulai melandai, bukan hanya dari SSE paling kecil.


```python
K = range(1, 11)
sse = []
for k in K:
    model = KMeans(n_clusters=k, random_state=0, n_init=10)
    model.fit(X)
    sse.append(model.inertia_)

hasil_elbow = pd.DataFrame({'k': list(K), 'SSE': sse})
display(hasil_elbow)
plt.figure(figsize=(8, 4))
plt.plot(list(K), sse, 'bx-')
plt.xticks(list(K))
plt.xlabel('Jumlah klaster (k)')
plt.ylabel('SSE')
plt.title('Metode Elbow pada Data Iris')
plt.grid(alpha=0.3)
plt.show()
```

## 6. Mencoba k = 3
Grafik Elbow menunjukkan penurunan besar sampai sekitar k = 3, kemudian penurunannya lebih kecil. Kita mencoba tiga klaster sebagai pilihan dari grafik tersebut. Nomor klaster tidak otomatis sama dengan nama spesies.


```python
model3 = KMeans(n_clusters=3, random_state=0, n_init=10)
labels3 = model3.fit_predict(X)
plt.figure(figsize=(7, 5))
plt.scatter(X.iloc[:, 0], X.iloc[:, 1], c=labels3, cmap='viridis', s=50)
plt.scatter(model3.cluster_centers_[:, 0], model3.cluster_centers_[:, 1],
            c='red', marker='X', s=200, label='Centroid')
plt.xlabel('Panjang sepal (cm)')
plt.ylabel('Lebar sepal (cm)')
plt.title('K-Means Iris dengan k = 3')
plt.legend()
plt.show()
print(f'SSE k=2: {cl_kmeans.inertia_:.3f}; SSE k=3: {model3.inertia_:.3f}')
```

## Kesimpulan
K-Means dapat mengelompokkan bunga berdasarkan kemiripan ukurannya. Jumlah klaster memengaruhi hasil dan SSE. Metode Elbow membantu memilih jumlah klaster tanpa memakai label spesies untuk melatih model.
