## 1. Import Library

```python
%matplotlib inline
# import library
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
import seaborn as sns
```

## 2. Langkah 1 — Membuat Data Dummy

Data dummy dengan dua kelas yang sudah terpisah dibuat menggunakan `make_blobs`.

```python
from sklearn.datasets import make_blobs

X, y = make_blobs(n_samples=50, centers=2,
                  random_state=0, cluster_std=0.60)
plt.scatter(X[:, 0], X[:, 1], c=y, s=50, cmap='autumn')
plt.show()
```

## 3. Langkah 2 — Ilustrasi Garis Pemisah

Dua himpunan data yang terbentuk sebenarnya dapat dipisahkan oleh lebih dari satu garis. Banyaknya garis pemisah berarti terdapat beberapa kemungkinan kelas untuk sebuah data baru, sehingga diperlukan kriteria untuk memilih garis yang paling baik.

## 4. Langkah 3 — Ilustrasi Margin

Margin dipakai untuk memperjelas posisi sebuah data terhadap label. Garis yang mampu memaksimalkan margin adalah garis pemisah yang paling optimal.

```python
xfit = np.linspace(-1, 3.5)
plt.scatter(X[:, 0], X[:, 1], c=y, s=50, cmap='autumn')

for m, b, d in [(1, 0.65, 0.33), (0.5, 1.6, 0.55), (-0.2, 2.9, 0.2)]:
    yfit = m * xfit + b
    plt.plot(xfit, yfit, '-k')
    plt.fill_between(xfit, yfit - d, yfit + d, edgecolor='none',
                     color='#AAAAAA', alpha=0.4)

plt.xlim(-1, 3.5)
plt.show()
```

## 5. Langkah 4 — Fitting Model

Model SVM dilatih dengan kernel linier. Selanjutnya dibuat fungsi bantu `plot_svc_decision_function` untuk memvisualisasikan batas keputusan dan margin.

```python
from sklearn.svm import SVC  # "Support vector classifier"

model = SVC(kernel='linear', C=1E10)
model.fit(X, y)
```

```python
# buat sebuah fungsi untuk menampilkan fitting data
def plot_svc_decision_function(model, ax=None, plot_support=True):
    if ax is None:
        ax = plt.gca()
    xlim = ax.get_xlim()
    ylim = ax.get_ylim()

    # buat grid untuk evaluasi model
    x = np.linspace(xlim[0], xlim[1], 30)
    y = np.linspace(ylim[0], ylim[1], 30)
    Y, X = np.meshgrid(y, x)
    xy = np.vstack([X.ravel(), Y.ravel()]).T
    P = model.decision_function(xy).reshape(X.shape)

    # plot batas dan margin
    ax.contour(X, Y, P, colors='k',
               levels=[-1, 0, 1], alpha=0.5,
               linestyles=['--', '-', '--'])

    # plot support vectors
    if plot_support:
        ax.scatter(model.support_vectors_[:, 0],
                   model.support_vectors_[:, 1],
                   s=300, linewidth=1, facecolors='none')
    ax.set_xlim(xlim)
    ax.set_ylim(ylim)

plt.scatter(X[:, 0], X[:, 1], c=y, s=50, cmap='autumn')
plot_svc_decision_function(model)
plt.show()
```

Titik data yang menyentuh area margin disebut **support vector**. Berikut data yang menjadi support vector.

```python
model.support_vectors_
```

## 6. Perbandingan Jumlah Data terhadap Model

Support vector menjadi kunci keberhasilan classifier. Data yang jauh dari margin tidak mengubah posisi garis pemisah. Hal ini diperlihatkan dengan membandingkan model yang dilatih pada 60 dan 120 data awal.

```python
def plot_svm(N=10, ax=None):
    X, y = make_blobs(n_samples=200, centers=2,
                      random_state=0, cluster_std=0.60)
    X = X[:N]
    y = y[:N]
    model = SVC(kernel='linear', C=1E10)
    model.fit(X, y)

    ax = ax or plt.gca()
    ax.scatter(X[:, 0], X[:, 1], c=y, s=50, cmap='autumn')
    ax.set_xlim(-1, 4)
    ax.set_ylim(-1, 6)
    plot_svc_decision_function(model, ax)

fig, ax = plt.subplots(1, 2, figsize=(16, 6))
fig.subplots_adjust(left=0.0625, right=0.95, wspace=0.1)
for axi, N in zip(ax, [60, 120]):
    plot_svm(N, axi)
    axi.set_title('N = {0}'.format(N))
plt.show()
```

Perubahan jumlah data tidak mengubah garis pemisah karena hanya support vector yang menentukan posisi hyperplane. Sel interaktif `interact` pada modul diganti dengan perbandingan statis N = 10 dan N = 200 berikut.

```python
# Perbandingan statis N = 10 dan N = 200 (pengganti sel interaktif ipywidgets)
fig, ax = plt.subplots(1, 2, figsize=(16, 6))
fig.subplots_adjust(left=0.0625, right=0.95, wspace=0.1)
for axi, N in zip(ax, [10, 200]):
    plot_svm(N, axi)
    axi.set_title('N = {0}'.format(N))
plt.show()
```
