# Pertemuan 11 — Tugas 1: Klasifikasi Suara dengan SVM (voice.csv)

Tugas ini membangun model SVM untuk mengklasifikasikan suara `male` dan `female` pada dataset `voice.csv`.

Ketentuan:
1. Split data dengan rasio 70:30 dan 80:20 untuk setiap model.
2. Gunakan kernel linier, polynomial, dan RBF.
3. Tabulasikan performansi setiap split dan kernel berdasarkan metrik akurasi.

Sumber: [Tugas Praktikum](https://polinema.gitbook.io/jti-modul-praktikum-pembelajaran-mesin-2025-2026/js11-klasifikasi-2/tugas-praktikum).

## 1. Import Library

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score
```

## 2. Load Data

```python
df = pd.read_csv('voice.csv')
print('Ukuran data:', df.shape)
print(df['label'].value_counts())
df.head()
```

## 3. Encoding Label

```python
df['label'] = df['label'].map({'male': 1, 'female': 0})
print(df['label'].value_counts())
```

## 4. Pisahkan Fitur dan Label

```python
X = df.drop(columns=['label'])
y = df['label']
print('Fitur:', X.shape)
print('Label:', y.shape)
```

## 5. Evaluasi Semua Kombinasi Split dan Kernel

Untuk setiap rasio split (70:30 dan 80:20) fitur distandardisasi lalu dibangun model SVM dengan kernel linier, polynomial, dan RBF. Akurasi data uji dicatat.

```python
hasil = []

for test_size, nama_split in [(0.30, '70:30'), (0.20, '80:20')]:
    X_train, X_test, y_train, y_test = train_test_split(
        X, y, test_size=test_size, random_state=42, stratify=y)

    scaler = StandardScaler()
    X_train = scaler.fit_transform(X_train)
    X_test = scaler.transform(X_test)

    for kernel in ['linear', 'poly', 'rbf']:
        model = SVC(kernel=kernel)
        model.fit(X_train, y_train)
        pred = model.predict(X_test)
        acc = accuracy_score(y_test, pred)
        hasil.append([nama_split, kernel, acc])

hasil_df = pd.DataFrame(hasil, columns=['Split', 'Kernel', 'Akurasi'])
hasil_df
```

## 6. Tabulasi Performansi

Tabel pivot menampilkan akurasi setiap kernel pada masing-masing rasio split.

```python
pivot = hasil_df.pivot(index='Kernel', columns='Split', values='Akurasi')
pivot
```
